## Exercise 4: Searching a list of keywords

A single rule in a rule-based chatbot owns several keywords. The rule applies when any one
of them is present.

**(a)** Using a `for` loop, a conditional statement and a boolean variable named `found`, determine
whether any element of `keywords` appears in `tokens`. Print `keyword found` or `no keyword found`
accordingly.

**(b)** Extend the code so that the matching keyword is also printed. Where several keywords match,
only the first is reported.

*Hint:* initialise `found = False` before the loop and set it to `True` inside the loop when a match
occurs. The `break` statement exits a loop immediately once the answer is known. Setting `found`
to `False` inside the loop is a common error: it causes every keyword after the first to overwrite
the result.

````python
# Exercise 4(a)

found = False

for keyword in keywords:
    if keyword in tokens:
        found = True
        break

if found:
    print("keyword found")
else:
    print("no keyword found")


# Exercise 4(b)

found = False
matching_keyword = None

for keyword in keywords:
    if keyword in tokens:
        found = True
        matching_keyword = keyword
        break

if found:
    print("keyword found")
    print("matching keyword:", matching_keyword)
else:
    print("no keyword found")
````

## Exercise 5: Scoring every rule

A chatbot holds many rules, not one. The dictionary `rules` maps each rule name, termed its
**tag**, to that rule's list of keywords. Determining which rule applies requires every rule to be
examined.

**(a)** Using nested `for` loops and a counter variable, compute for each tag the number of its
keywords that appear in `tokens`. Store the results in a dictionary named `scores`, with the tag as
the key and the count as the value. Print `scores`.

**(b)** Using a `for` loop over `scores` and two tracking variables, determine the tag with the
highest score. Print the tag and its score. Where two tags hold the same highest score, the tag
encountered first is retained.

**(c)** Extend the code so that `no match` is printed when the highest score is 0.

*Hint:* iterate over a dictionary with `for tag, keyword_list in rules.items():`, which yields the
key and value together. For part (b), initialise `best_tag = None` and `best_score = 0`, and replace
them only when a strictly greater score is found — the comparison `>` rather than `>=` is what
causes the first of two equal scores to be retained.

````python
# Exercise 5(a)

scores = {}

for tag, keyword_list in rules.items():
    count = 0

    for keyword in keyword_list:
        if keyword in tokens:
            count += 1

    scores[tag] = count

print(scores)


# Exercise 5(b)

best_tag = None
best_score = 0

for tag, score in scores.items():
    if score > best_score:
        best_score = score
        best_tag = tag

print("best tag:", best_tag, "with score", best_score)
````

## Exercise 6: Encapsulating the text preparation

The code written in Exercises 1 to 3 operates on one fixed message. A function allows the
same procedure to be applied to any message without the code being rewritten.

**(a)** Define a function `normalise(text)` that returns `text` in lower case, with leading and
trailing whitespace removed and with all characters in `PUNCTUATION` removed. The body reuses the
logic of Exercises 1 and 2.

**(b)** Define a function `tokenise(text)` that returns the list of tokens of `text`. The body calls
`normalise` and then splits the result.

**(c)** Apply `tokenise` to the three messages below and print each result.

```python
"What TIME do you open?"
"   HOW MUCH IS A LARGE??   "
"do you deliver to Remera"
```

*Hint:* the order in which the two operations are combined affects the result. Removing punctuation
from `"hours ?"` before stripping whitespace yields `"hours"`; stripping whitespace first yields
`"hours "`, with a trailing space that will produce an unwanted empty token. Both orders should be
tested and the correct one retained.

````python
def normalise(text):
    """Return text in lower case, without surrounding whitespace or punctuation."""

    text = text.lower().strip()

    cleaned = ""

    for ch in text:
        if ch not in PUNCTUATION:
            cleaned += ch

    return cleaned


def tokenise(text):
    """Return the list of tokens of text."""

    return normalise(text).split()
messages = [
    "What TIME do you open?",
    "   HOW MUCH IS A LARGE??   ",
    "do you deliver to Remera"
]
for text in messages:
    print(tokenise(text))
````

## Exercise 7: Parsing rules and scoring

Holding rules inside program code requires a programmer for every change. Rules are more
usefully held as text, which may be stored in a separate file and edited without altering the
program.

**(a)** Define a function `parse_rules(text)` that converts the block `RULES_TEXT` below into a
dictionary mapping each tag to its list of keywords, producing a value equivalent to the `rules`
dictionary used in Exercise 5.

**(b)** Define a function `score_all(tokens, rules)` that returns a dictionary of tag to score,
reusing the logic of Exercise 5(a).

**(c)** Define a function `best_match(scores, threshold=1)` that returns the highest-scoring tag, or
`None` when the highest score is below `threshold`. The parameter `threshold` takes the value 1 when
not supplied by the caller.

*Hint:* `text.strip().splitlines()` yields the non-empty lines as a list. `line.split("|")` divides a
line into the tag and the keyword portion, and `.split(",")` divides that portion into individual
keywords. Every piece carries surrounding whitespace and must be passed through `.strip()`; omitting
this produces keywords such as `" open"`, which never match.

````python
RULES_TEXT = """
hours    | hour, hours, open, close, time
delivery | deliver, delivery, bring, send
price    | price, cost, much, rwf
menu     | menu, options, have
order    | order, buy, want
"""
````

````python
# Exercise 7(a)
def parse_rules(text):
    """Convert the rules text into a dictionary of tag to keyword list."""
    parsed = {}

    lines = text.strip().splitlines()

    for line in lines:
        parts = line.split("|")
        tag = parts[0].strip()
        keyword_text = parts[1].strip()

        keywords_list = []

        for keyword in keyword_text.split(","):
            keywords_list.append(keyword.strip())

        parsed[tag] = keywords_list

    return parsed


# Exercise 7(b)
def score_all(tokens, rules):
    """Return a dictionary of tag to score for these tokens."""
    scores = {}

    for tag, keyword_list in rules.items():
        score = 0

        for keyword in keyword_list:
            if keyword in tokens:
                score += 1

        scores[tag] = score

    return scores


# Exercise 7(c)
def best_match(scores, threshold=1):
    """Return the highest-scoring tag, or None if the highest score is below threshold."""
    best_tag = None
    best_score = 0

    for tag, score in scores.items():
        if score > best_score:
            best_score = score
            best_tag = tag

    if best_score < threshold:
        return None

    return best_tag

parsed_rules = parse_rules(RULES_TEXT)
print(parsed_rules)
````


## Exercise 8: Assembling the matching procedure

The functions defined in Exercises 6 and 7 are now combined into a single procedure. No new
logic is required.

**(a)** Define a function `match(message, rules)` that returns a tuple `(tag, score)` for the
best-matching rule, and the tuple `("fallback", 0)` when no rule reaches the threshold.

**(b)** Apply `match` to each message in `test_messages` below and print a table showing the message,
the tag returned and the score.

```python
test_messages = [
    "What time do you open?",
    "do you deliver to Kicukiro",
    "HOW MUCH IS A LARGE??",
    "can I see the menu",
    "I want to order",
    "do u delivr",
    "asdfgh",
    "",
]
```

*Hint:* `best_match` returns `None` when nothing reaches the threshold; that value indicates the
fallback case. A function returns several values at once by separating them with a comma:
`return tag, scores[tag]`. Field widths in an f-string, written as `{message!r:30}`, produce aligned
columns.

````python
test_messages = [
    "What time do you open?",
    "do you deliver to Kicukiro",
    "HOW MUCH IS A LARGE??",
    "can I see the menu",
    "I want to order",
    "do u delivr",
    "asdfgh",
    "",
]


# Exercise 8(a)
def match(message, rules):
    """Return (tag, score) for the best-matching rule, or ("fallback", 0)."""
    tokens = tokenise(message)
    scores = score_all(tokens, rules)
    tag = best_match(scores)

    if tag is None:
        return "fallback", 0

    return tag, scores[tag]


# Exercise 8(b)
print(f"{'message':30} {'tag':12} {'score'}")
print("-" * 52)

for message in test_messages:
    tag, score = match(message, parsed_rules)
    print(f"{message!r:30} {tag:12} {score}")
````

## Exercise 9: Interactive matching (Bonus Question)

Write a program that repeatedly accepts a message from the user, applies `match` to it, and prints
the tag and score returned. The program terminates when the user enters `bye`.

The program must additionally:

- print an opening line stating which tags the program recognises;
- record every message that returns `fallback` in a list named `unmatched`;
- print the contents of `unmatched` on termination.

*Hint:* use a `while True:` loop with `input()`, and `break` when the entered message equals `bye`.
Test the termination condition before applying `match`, so that `bye` is not treated as a message to
be matched. Compare the entered value for equality with `bye` rather than using the `in` operator: a
message such as `"thanks, what are your opening hours"` contains the substring `bye` in no form, but
a keyword list containing `thanks` would terminate the program if `in` were used.

*Note:* this cell requires keyboard input and will wait indefinitely if the notebook is run with
**Run all**. Execute it individually.

**Sample interaction**

```
Recognised topics: hours, delivery, price, menu, order
Enter a message, or 'bye' to finish.

> what time do you open
    hours (score 2)

> do u delivr
    fallback (score 0)

> bye

Unmatched messages recorded: 1
  1. do u delivr
```

````python
# Exercise 9 (Bonus) - run this cell individually

unmatched = []

print("Recognised topics:", ", ".join(parsed_rules.keys()))
print("Enter a message, or 'bye' to finish.")

while True:
    message = input("> ")

    if message == "bye":
        break

    tag, score = match(message, parsed_rules)

    print(f"    {tag} (score {score})")

    if tag == "fallback":
        unmatched.append(message)

print()
print("Unmatched messages recorded:", len(unmatched))

for i, message in enumerate(unmatched, start=1):
    print(f"  {i}. {message}")
````
