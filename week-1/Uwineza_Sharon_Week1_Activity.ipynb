```python
print("🍕 Welcome to the sharon's Pizza!")

while True:

    size = input("What size pizza would you like? (Small, Medium, Large): ")

    if size.lower() == "bye":
        print("Goodbye! 👋")
        break

    topping = input("What topping would you like? (Pepperoni, Mushrooms, Cheese): ")

    if topping.lower() == "bye":
        print("Goodbye! 👋")
        break

    drink = input("Would you like a drink? (Orange, Coca-Cola, Fanta, Citro, No): ")

    if drink.lower() == "bye":
        print("Goodbye! 👋")
        break

    if size.lower() == "small":
        price = 9000
    elif size.lower() == "medium":
        price = 12000
    elif size.lower() == "large":
        price = 15000
    else:
        print("Sorry, that is not a valid pizza size. Let's start again.\n")
        continue

    if drink.lower() in ["orange", "coca-cola", "fanta", "citro"]:
        price += 1000
        drink_message = f" with a {drink.title()}"

    elif drink.lower() == "no":
        drink_message = ""

    else:
        print("Sorry, that is not a valid drink. Let's start again.\n")
        continue

    print(f"\nYes! Your {topping.title()} {size.title()} pizza{drink_message}.")
    print(f"Your total is {price:,} RWF.")

    confirm = input("Would you like to confirm your order? (Yes/No): ")

    if confirm.lower() == "bye":
        print("Goodbye! 👋")
        break

    if confirm.lower() == "yes":
        print("\nAwesome! Your order is confirmed.")
        print("Please wait 15 minutes.")
        print("Bye! 👋")
        break

    elif confirm.lower() == "no":
        print("\nNo problem! Let's start your order again.\n")
        continue

    else:
        print("\nPlease answer Yes or No. Let's start your order again.\n")
        continue
```
